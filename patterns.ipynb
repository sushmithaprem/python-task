Patterns

# 1.Right Triangle

```python
for i in range(1,6):
    for j in range(i):
        print("*" ,end= " ")
    print()
    
```

# 2. Square Pattern

```python
for i in range(4):
    for j in range(4):
        print("*", end=" ")
    print()
```

# 3. Number Triangle

```python
for i in range(1,6):
    for j in range(1,i+1):
        print(j, end=" ")
    print()
```

# 4. Inverted Triangle

```python
for i in range(5,0,-1):
    print("* " * i)
```

# 5. Pyramid

```python
for i in range(1,6):
    # for space 
    for j in range(5 - i):
        print(" ",end ="")
    # for star
    for j in range(i):
        print("*", end=" ")
    print()
```

# 6. Inverse Pyramid

```python
for i in range(5):
    for j in range(i):
        print(" ",end="")
    for j in range(5-i):
        print("*",end=" ")
    print()
```

# 7.Diamond Pattern

```python
n = 5
# upper 

for i in range(1,n+1):
    for j in range(n-i):
        print(" ",end="")
    for j in range(2*i-1):
        print("*",end="")
    print()

# Lower 

for i in range (n-1,0,-1):
    for j in range(n-i):
        print(" ", end="")
    for j in range (2 * i -1):
        print("*",end="")
    print()
```

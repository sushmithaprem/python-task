# 8.Amstrong Number

```python
num = int(input("Enter a number: "))
temp = num
count = 0

while temp > 0:
    digit = temp % 10
    count += digit ** 3
    temp //= 10

if count == num:
    print(num,"Amstrong Number....")
else:
    print(num,"Not an Amstrong Number....")
```

# 9.Check whether a number is a palindrome

```python
num = int(input())
temp = num
rev = 0

while temp > 0:
    digit = temp % 10
    rev = rev * 10 + digit
    temp = temp//10

if rev == num:
    print(num,"...Its an Palindrom")
else:
    print(num,"...Its Not an Palindrom")
```

# 13.Find GCD of two numbers

```python
n1 = 10
n2 = 40

while n2 != 0:
    rem = n1 % n2
    n1 = n2
    n2 = rem

print(n1)
```

# 14. Find LCM of two numbers

```python
n1 = 12
n2 = 18

a = n1
b = n2

while b != 0:
    rem = a % b
    a = b
    b = rem

gcd = a

lcm = (n1 * n2) // gcd
print(gcd)
print(lcm)

```

# remove the duplicates

```python
def dupli(l):
  
  s = set(l)
  return s


l = [1,2,2,3,4,4,5]
result = dupli(l)
print(result)
```

# Find duplicate elements

```python
def dup(l):
    seen = set()
    duplicate = []
    for i in l:
        if i in seen:
            duplicate.append(i)
        else:
            seen.add(i)
    return duplicate
l = [1,2,2,3,4,4,5]
result  = dup(l)
print(result)
```

# Reverse a list

```python
def rev(l):
    
    s = l[::-1]
    return s


l = [1,2,2,3,4,4,5]
result  = rev(l)
print(result)
```

# Move zeros to end

```python
def zero(l):
    result = []
    count = 0
    for i in l:
        if i != 0:
           result.append(i) 
        else:
            count = count + 1
    for i in range(count):
        result.append(0)            

    return result



l = [0,1,5,0,8,0,5]
result = zero(l)
print(result)

```

# rotate an list

```python
def rotate(l,k):
    k = k % len(l)
    result = l[-k:] + l[:-k]
    return result

l = [1,2,3,4,5]
k = 1

result = rotate(l,k)
print(result)
```

# Character frequency

```python
def freq(s):
    count = {}
    s = s.lower()
    for i in s:
        if i in count:
            count[i] += 1
        else:
            count[i] = 1
    return count 


s = input()
result = freq(s)
print(result )
```

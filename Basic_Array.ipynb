3. Find the largest element

```python
arr = [10,60,70,100,20]
large = 0
for i in arr:
    if i > large:
        large = i
print(large)
```

4. Find the smallest element

5. Count even numbers

```python
arr = [10,3,70,100,20]
count = 0
for i in arr:
    if i% 2 ==0:
        print(i)
        count = count+1
print("Count of even numbers:",count)
        
```

```python
n = int(input("Enter a size: "))
arr = list(map(int, input("Enter a elements: ").split()))
count = 0
for i in arr:
    if i% 2 ==0:
        print(i)
        count = count+1
print("Count of even numbers:",count)
```

6. Count odd numbers

```python
arr = [10,3,70,100,20]
count = 0
for i in arr:
    if i% 2 !=0:
        print(i)
        count = count+1
print("Count of odd numbers:",count)
        
```

```python
n = int(input("Enter a size: "))
arr = list(map(int, input("Enter a elements: ").split()))
count = 0
for i in arr:
    if i% 2 !=0:
        print(i)
        count = count+1
print("Count of odd numbers:",count)
```

7. Find the average

```python
a = [10,20,30,40,50]
print(sum(a))
avg = sum(a)/len(a)
print("Average:", avg)
```

# 1.Print each character of a string

```python
name = "python"
for i in name:
    print(i)
```

# 2.Find the length of a string

```python
s = "Sushmitha"
print(len(s))
```

# 3.Count the number of vowels

```python
str = "education"
vowels = ["a","e","i","o","u"]
count = 0
for char in str:
    if char in vowels:
        count = count + 1
print("Number of vowels:", count)
      
```

# 4. Count vowels and consonants

```python
str1 = "python"
vowels = ["a","e","i","o","u"]
count = 0
rem = 0
for char in str1:
    if char in vowels:
        count = count + 1
    else:
        rem = rem + 1
print("Vowels count:" ,count)
print("Consonant count:" ,rem)
```

# 5.Count spaces in a string

```python
text = " hello world python "
count = 0
for char in text:
    if char == " ":
        count = count +1
print("Number of spaces:", count)
```

# 6.Convert lowercase to uppercase

```python
txt = "python learning"
print(txt.upper())
```

# 7.Convert uppercase to lowercase 

```python
txt = "PYTHON LEARNING"
print(txt.lower())
```

# 8.Count occurrence of a character

```python
text = "banana"
target = "n"

count = 0
for char in text:
    if char == target:
        count = count + 1
print(target, "occurs", count, "times")
```

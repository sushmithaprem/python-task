# 3.Count the number of vowels

```python
str = "education"
vowels = ["a","e","i","o","u"]
count = 0
for char in str:
    if char in vowels:
        count = count + 1
print("Number of vowels:", count)
      
```

# 4. Count vowels and consonants

```python
str1 = "python"
vowels = ["a","e","i","o","u"]
count = 0
rem = 0
for char in str1:
    if char in vowels:
        count = count + 1
    else:
        rem = rem + 1
print("Vowels count:" ,count)
print("Consonant count:" ,rem)
```

# 5.Count spaces in a string

```python
text = " hello world python "
count = 0
for char in text:
    if char == " ":
        count = count +1
print("Number of spaces:", count)
```

# 8.Count occurrence of a character

```python
text = "banana"
target = "n"

count = 0
for char in text:
    if char == target:
        count = count + 1
print(target, "occurs", count, "times")
```

# 10.Check whether a string is palindrome

```python
#str = "appa"
str = input("Enter a word: ")
print(str)
rev = str[::-1]
if str == rev:
    print("Palindrom")
else:
    print("Not Palindrom")
```

# 13.Print characters at even positions

```python
str = "sushmitha"

for i in range(len(str)):
    if i % 2 == 0 :
        print(str[i])
```

# 14.Print characters at odd positions

```python
str = "sushmitha"

for i in range(len(str)):
    if i % 2 != 0 :
        print(str[i])
```

# 15.Count digits in a string

```python
num = "abc12python345"
count = 0
for char in num:
    if char.isdigit():
        count = count + 1
print("Number of digits:", count)
```

# 16. Count alphabets in a string

```python
num = "abc12python345"
count = 0
for char in num:
    if char.isalpha():
        count = count + 1
print("Number of digits:", count)
```

# 17.Count special characters

```python
num = "abc12!python@345="
count = 0
for char in num:
    if not char.isalpha() and not char.isdigit():
        count = count + 1
print("Number of digits:", count)
```

# 18.Find the largest character
(using ascii key values based)

```python
text = "python"

largest = text[0]

for char in text:
    if char > largest:
        largest = char
print("Largest character:", largest)
```
